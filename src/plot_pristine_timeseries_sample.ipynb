```python
import seaborn

from matplotlib import pyplot

from pristine import load_pristine_timeseries
from self_lensing import gen_sls, default_times
```

```python
a_limit = 5
```

```python
pts = load_pristine_timeseries()
```

```python
for _, ts in pts[pts['a'] < a_limit].sample(n=10).iterrows():
    sls = gen_sls(a=ts['a'], e=ts['e'], periapsis_phase=ts['periapsis_phase'], macc=ts['Macc'])
    print('a=', ts['a'], '; e=', ts['e'], '; periapsis phase=', ts['periapsis_phase'], '; Porb=', sls.porb, '; Macc=', ts['Macc'])
    seaborn.lineplot({'times': default_times, 'rv': ts['rv']}, x='times', y='rv')
    pyplot.show()
    pyplot.close()
```
