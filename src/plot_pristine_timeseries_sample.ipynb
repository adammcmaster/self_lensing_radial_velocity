```python
import numpy
import seaborn
from matplotlib import pyplot

from pristine import load_pristine_rv_curve, load_pristine_timeseries
```

```python
a_limit = 150
sample_size = 10
```

```python
pts = load_pristine_timeseries()
pts.head()
```

```python
sample = pts[pts['a'] < a_limit]
if len(sample) > sample_size:
    sample = sample.sample(n=sample_size, random_state=1)

for _, ts in sample.iterrows():
    curve = load_pristine_rv_curve(ts)
    times = curve['rv_times']
    rv = curve['rv']
    period = ts['Porb']

    print(ts[['simulation_id', 'a', 'e', 'periapsis_phase', 'Macc', 'Mcomp', 'Rcomp', 'Porb', 'tau_eff', 'rv_path']])

    seaborn.lineplot(x=times, y=rv, c='red')
    seaborn.scatterplot(x=times, y=rv)
    pyplot.xlabel('JD')
    pyplot.ylabel('RV [m / s]')
    pyplot.show()
    pyplot.close()

    folded_times = (times - ts['peak_time']) % period
    seaborn.lineplot(x=folded_times, y=rv, c='red')
    seaborn.scatterplot(x=folded_times, y=rv)
    pyplot.xlabel('Phase time [d]')
    pyplot.ylabel('RV [m / s]')
    pyplot.show()
    pyplot.close()
```
