```python
import numpy
import seaborn

from matplotlib import pyplot

from pristine import load_pristine_timeseries, load_pristine_lombscargle
from self_lensing import default_times
```

```python
a_limit = 5
```

```python
pts = load_pristine_timeseries()
pls = load_pristine_lombscargle()
```

```python
for _, ts in pts[pts['a'] < a_limit].sample(n=10).iterrows():
    print(ts)
    period = pls[pls['pristine_idx'] == ts.name]['top_period'].iloc[0]
    seaborn.lineplot({'times': default_times, 'rv': ts['rv']}, x='times', y='rv', c='red')
    seaborn.scatterplot({'times': default_times, 'rv': ts['rv']}, x='times', y='rv')
    pyplot.show()
    pyplot.close()

    print('period=', period)

    seaborn.lineplot({'times': numpy.array(default_times) % period, 'rv': ts['rv']}, x='times', y='rv', c='red')
    _ts = ts.to_dict()
    _ts['times'] = numpy.array(default_times) % period
    seaborn.scatterplot(_ts, x='times', y='rv')
    pyplot.show()
    pyplot.close()
```
