```python
import numpy
import pandas
import seaborn

from astropy import units 
from astropy.time import Time

from matplotlib import pyplot

from selflensing.generators import RVGenerator
from selflensing.models import SelfLensingSystem

from pristine import load_pristine_timeseries

from fitting import TIMESERIES_TABLE_PATH
from utils import DEFAULTS
```

```python
pts = load_pristine_timeseries()
```

```python
pyplot.rcParams["figure.figsize"] = [12.8, 9.6]
```

```python
results = pandas.read_parquet(TIMESERIES_TABLE_PATH)
results = results.sample(10)
```

```python
results['close'] = (
    numpy.isclose(results['a'], results['fitted_a'], rtol=0.1)
    & numpy.isclose(results['Macc'], results['fitted_Macc'], rtol=0.1)
    & numpy.isclose(results['Porb'], results['fitted_porb'], rtol=0.1)
)
results['close'].value_counts()
```

```python
for idx, row in results.iterrows():
    print(row)
    sls = SelfLensingSystem(
        a=row['fitted_a'] * units.au,
        b=DEFAULTS["b"],
        Macc=row['fitted_Macc'] * units.M_sun,
        e=row['fitted_e'] * units.dimensionless_unscaled,
        periapsis_phase=row['fitted_periapsis_phase'] * units.rad,
        Mcomp=DEFAULTS["Mcomp"],
        Rcomp=DEFAULTS["Rcomp"],
        Teff=DEFAULTS["Teff"],
    )
    rvg = RVGenerator(sls, times=Time(row['rv_times'], format='jd'), peak_time=Time(row['peak_time'], format='jd'))

    seaborn.scatterplot(row.to_dict(), x='rv_times', y='rv')
    seaborn.lineplot({'times': row['rv_times'], 'rv': rvg.rv}, x='times', y='rv', color='red')
    pyplot.show()
```
